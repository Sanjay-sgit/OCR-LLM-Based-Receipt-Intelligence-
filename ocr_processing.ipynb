```python
from google.colab import files
from pathlib import Path
import zipfile
import time
import csv
import os
import paddle
from paddleocr import PaddleOCR

uploaded = files.upload()
zip_path = next(
    name for name in uploaded
    if name.lower().endswith(".zip")
)

IMAGE_DIR = Path("/content/receipts")
IMAGE_DIR.mkdir(parents=True, exist_ok=True)

with zipfile.ZipFile(zip_path) as zf:
    zf.extractall(IMAGE_DIR)

images = sorted(
    p for p in IMAGE_DIR.rglob("*")
    if p.is_file()
    and p.suffix.lower() in {".jpg", ".jpeg", ".png"}
)

print("Images found:", len(images))
print("GPU enabled:", paddle.device.is_compiled_with_cuda())

device = "gpu" if paddle.device.is_compiled_with_cuda() else "cpu"

ocr = PaddleOCR(
    lang="en",
    device=device,
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
)

print("OCR model initialized on:", device)
```

```python
for path in images[:30]:
    print("\nFILE:", path.name)

    prediction = ocr.predict(str(path))

    for result in prediction:
        data = getattr(result, "json", None)
        if callable(data):
            data = data()

        if not isinstance(data, dict):
            continue

        data = data.get("res", data)

        for text in data.get("rec_texts", []):
            print(text)
```

```python
def get_text(prediction):
    lines = []

    for result in prediction:
        # PaddleOCR v3 result format
        data = getattr(result, "json", None)

        if callable(data):
            data = data()

        if isinstance(data, dict):
            data = data.get("res", data)

            texts = data.get("rec_texts", [])
            if texts:
                lines.extend(str(t) for t in texts if t)
                continue

        # Compatibility with older PaddleOCR result formats
        if isinstance(result, list):
            for line in result:
                if len(line) >= 2 and len(line[1]) >= 1:
                    lines.append(str(line[1][0]))

    return "\n".join(lines)
```

```python
import time
from pathlib import Path

# Confirm that the OCR model and image list exist
print("Total images:", len(images))

sample = images[:30]
success = 0
failed = 0
total_text_chars = 0

start = time.perf_counter()

for i, path in enumerate(sample, start=1):
    try:
        prediction = ocr.predict(str(path))
        text = get_text(prediction)

        if text.strip():
            success += 1
            total_text_chars += len(text)
        else:
            failed += 1

    except Exception as e:
        failed += 1
        print(f"Error on {path.name}: {str(e)[:150]}")

    elapsed = time.perf_counter() - start
    print(
        f"{i}/30 | Success: {success} | Failed: {failed} | "
        f"Elapsed: {elapsed:.1f}s"
    )

elapsed = time.perf_counter() - start

print("\n--- OCR BENCHMARK ---")
print("Images tested:", len(sample))
print("Successful:", success)
print("Failed:", failed)
print("Text characters:", total_text_chars)
print(f"Total time: {elapsed:.1f} seconds")
print(f"Average per image: {elapsed / len(sample):.2f} seconds")
print(f"Estimated OCR time for 626 images: {elapsed * 626 / len(sample) / 60:.1f} minutes")
```

```python
import pandas as pd

OUTPUT = Path("/content/ocr_results.csv")
CHECKPOINT = Path("/content/ocr_checkpoint.csv")

results = {}
if CHECKPOINT.exists():
    previous = pd.read_csv(CHECKPOINT, keep_default_na=False)
    for _, row in previous.iterrows():
        if row["status"] == "ok":
            results[row["filename"]] = row["raw_ocr_text"]

def get_text(prediction):
    lines = []
    for result in prediction:
        data = getattr(result, "json", None)
        if callable(data):
            data = data()
        if not isinstance(data, dict):
            continue
        data = data.get("res", data)
        lines.extend(data.get("rec_texts", []))
    return "\n".join(str(t) for t in lines if t)

failed = {}
start = time.perf_counter()

for i, path in enumerate(images, start=1):
    if path.name in results:
        continue

    try:
        text = get_text(ocr.predict(str(path)))
        if text.strip():
            results[path.name] = text
        else:
            failed[path.name] = "No text recognized"
    except Exception as e:
        failed[path.name] = str(e)[:500]

    if i % 5 == 0 or i == len(images):
        rows = []
        for p in images:
            if p.name in results:
                rows.append({
                    "filename": p.name,
                    "raw_ocr_text": results[p.name],
                    "status": "ok",
                    "error": "",
                })
            elif p.name in failed:
                rows.append({
                    "filename": p.name,
                    "raw_ocr_text": "",
                    "status": "failed",
                    "error": failed[p.name],
                })

        pd.DataFrame(rows).to_csv(CHECKPOINT, index=False)

    print(
        f"{i}/{len(images)} | "
        f"success={len(results)} | failed={len(failed)} | "
        f"elapsed={(time.perf_counter()-start)/60:.1f} min"
    )

pd.DataFrame([
    {"filename": name, "raw_ocr_text": text}
    for name, text in results.items()
]).to_csv(OUTPUT, index=False, encoding="utf-8-sig")

print("OCR complete.")
print("Successful:", len(results))
print("Failed:", len(failed))
print("Minutes:", round((time.perf_counter()-start)/60, 2))
```
